# NASA Exoplanet Archive - Data Cleaning Pipeline
This Jupyter Notebook contains a robust Python data-cleaning pipeline developed for the Python + Power BI Data Visualization course.

The notebook performs setup, trimming, missing value analysis, two-stage imputation, and column renaming to output a pristine CSV (`cleaned_exoplanets.csv`) optimized for a dark-themed Power BI dashboard and aligned perfectly with lab report template requirements.

## 1. INITIAL SETUP & COLUMN TRIMMING
Load the exoplanet dataset, handle NASA's metadata comment lines starting with `#`, and filter immediately to the 29 core visualization columns.

```python
# ==========================================

# 1. INITIAL SETUP & COLUMN TRIMMING

# ==========================================

import os

import pandas as pd

import numpy as np



# Load the exoplanet dataset, skipping NASA's comment lines starting with '#'

df = pd.read_csv('nasa_exoplanets.csv', comment='#')



# Filter to retain only the 29 core visualization columns

core_columns = [

    'pl_name', 'hostname', 'sy_snum', 'sy_pnum', 'sy_mnum', 'discoverymethod', 

    'disc_year', 'disc_facility', 'pl_orbper', 'pl_orbsmax', 'pl_rade', 'pl_radj', 

    'pl_bmasse', 'pl_bmassj', 'pl_bmassprov', 'pl_orbeccen', 'pl_insol', 'pl_eqt', 

    'st_spectype', 'st_teff', 'st_rad', 'st_mass', 'st_met', 'st_metratio', 

    'st_logg', 'ra', 'dec', 'sy_dist', 'sy_gaiamag'

]

df = df[core_columns]

print("Raw dataset loaded and filtered successfully.")
```

## 3. REPORT SECTION 2.3: MISSING VALUE ANALYSIS
Generate a clean, copy-pasteable Markdown table matching your lab report layout that details null counts, missing percentage, and planned action strategy.

```python
# ==========================================

# 3. REPORT SECTION 2.3: MISSING VALUE ANALYSIS

# ==========================================

# Execute isnull().sum() to display null counts

null_counts = df.isnull().sum()

print("Missing values per raw column:")

print(null_counts[null_counts > 0])



print("\n" + "="*80)

print("REPORT SECTION 2.3: MISSING VALUE ANALYSIS MARKDOWN TABLE")

print("="*80)



# Generate a clean, text-based Markdown table matching the report layout

# Columns: | Column Name | Missing Count | % Missing | Action Taken |

print("| Column Name | Missing Count | % Missing | Action Taken |")

print("|---|---|---|---|")

total_rows = len(df)

for col in df.columns:

    missing_count = null_counts[col]

    if missing_count > 0:

        pct_missing = (missing_count / total_rows) * 100

        # Determine imputation strategy / Action Taken

        if df[col].dtype == 'object':

            action = "Filled with 'Unknown' for categorical text"

        else:

            action = "Imputed via group median based on discovery method; global fallback if needed"

        print(f"| {col} | {missing_count} | {pct_missing:.2f}% | {action} |")
```

## 4. REPORT SECTION 2.4: CLEANING ACTIONS & COLUMN RENAMING
Perform sequential cleaning actions: remove duplicates, apply two-stage imputation for numeric values, fill categorical nulls with "Unknown", force-cast `disc_year` to `int`, and rename all 29 columns. Outputs a copy-pasteable Markdown summary matching the Report Section 2.4 grid.

```python
# ==========================================

# 4. REPORT SECTION 2.4: CLEANING ACTIONS & COLUMN RENAMING

# ==========================================

# a) Remove duplicate rows

initial_rows = len(df)

df.drop_duplicates(inplace=True)

duplicates_removed = initial_rows - len(df)



# Identify numeric and categorical columns

numeric_cols = df.select_dtypes(include=['number']).columns.tolist()

categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()



# b) Handle missing values with a robust two-stage strategy

# 1. Fill numeric with group median based on 'discoverymethod'

for col in numeric_cols:

    group_medians = df.groupby('discoverymethod')[col].transform('median')

    df[col] = df[col].fillna(group_medians)

    

    # 2. Fill remaining numeric nulls with global median fallback

    global_median = df[col].median()

    if not pd.isna(global_median):

        df[col] = df[col].fillna(global_median)

    else:

        df[col] = df[col].fillna(0)



# 3. Fill categorical cells with "Unknown"

for col in categorical_cols:

    df[col] = df[col].fillna("Unknown")



# c) Force-cast disc_year to an integer type (int)

df['disc_year'] = df['disc_year'].astype(int)



# d) Convert Distance from Parsecs to Light Years (1 Parsec = 3.26156 Light Years)

df['sy_dist'] = df['sy_dist'] * 3.26156



# e) RENAME ALL 29 COLUMNS

rename_dict = {

    'pl_name': 'Planet_Name',

    'hostname': 'Host_Star_Name',

    'sy_snum': 'Number_of_Stars',

    'sy_pnum': 'Number_of_Planets',

    'sy_mnum': 'Number_of_Moons',

    'discoverymethod': 'Discovery_Method',

    'disc_year': 'Discovery_Year',

    'disc_facility': 'Discovery_Facility',

    'pl_orbper': 'Orbital_Period_Days',

    'pl_orbsmax': 'Orbital_Distance_AU',

    'pl_rade': 'Planet_Radius_Earth_Units',

    'pl_radj': 'Planet_Radius_Jupiter_Units',

    'pl_bmasse': 'Planet_Mass_Earth_Units',

    'pl_bmassj': 'Planet_Mass_Jupiter_Units',

    'pl_bmassprov': 'Mass_Measurement_Type',

    'pl_orbeccen': 'Orbit_Eccentricity',

    'pl_insol': 'Insolation_Flux_Earth_Units',

    'pl_eqt': 'Equilibrium_Temperature_K',

    'st_spectype': 'Star_Spectral_Type',

    'st_teff': 'Star_Temperature_K',

    'st_rad': 'Star_Radius_Solar_Units',

    'st_mass': 'Star_Mass_Solar_Units',

    'st_met': 'Star_Metallicity',

    'st_metratio': 'Star_Metallicity_Ratio',

    'st_logg': 'Star_Surface_Gravity',

    'ra': 'Right_Ascension_Deg',

    'dec': 'Declination_Deg',

    'sy_dist': 'Distance_From_Earth_Light_Years',

    'sy_gaiamag': 'Star_Gaia_Magnitude'

}

df.rename(columns=rename_dict, inplace=True)



# Generate and print Section 2.4 report grid table

print("="*80)

print("REPORT SECTION 2.4: CLEANING ACTIONS & RENAMING SUMMARY")

print("="*80)

print("| # | Action | Pandas Code Used | Result / Reason |")

print("|---|---|---|---|")

print(f"| 1 | Remove Duplicates | `df.drop_duplicates(inplace=True)` | Removed {duplicates_removed} duplicate rows (dataset is unique). |")

print("| 2 | Impute Numeric Columns (Stage 1) | `df[col] = df[col].fillna(df.groupby('discoverymethod')[col].transform('median'))` | Group-based median imputation using discovery method to preserve astronomical relationships. |")

print("| 3 | Impute Numeric Columns (Stage 2) | `df[col] = df[col].fillna(df[col].median())` | Global median fallback imputation applied to ensure zero remaining missing values. |")

print("| 4 | Fill Categorical Columns | `df[col] = df[col].fillna('Unknown')` | Missing categorical values in 'st_spectype' and 'st_metratio' filled with 'Unknown'. |")

print("| 5 | Type Casting | `df['disc_year'] = df['disc_year'].astype(int)` | Explicitly cast Discovery Year to integer for clean database styling and time-series plotting. |")

print("| 6 | Unit Conversion | `df['sy_dist'] = df['sy_dist'] * 3.26156` | Converted distance unit from Parsecs to Light Years (factor of 3.26156) for more intuitive public-facing visualization. |")

print("| 7 | Column Renaming | `df.rename(columns=rename_dict, inplace=True)` | Renamed all 29 fields to highly expressive, human-readable column headers for Power BI (e.g. Distance_From_Earth_Light_Years). |")
```
